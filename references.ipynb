# Mathematical References

<div style="height: 15px;"></div>

These sources describe the mathematical concepts used
in the Earth, Fire, and Aether experiments.

The methods were implemented directly in Python using NumPy.

### [1] Frame-Based Audio Signal Level
[librosa — RMS Energy](https://librosa.org/doc/0.11.0/generated/librosa.feature.rms.html)

Background for the calculation of signal level in short
audio windows. This project calculates mean-square amplitude.

### [2] Normalized Vector Similarity
[SciPy — Cosine Distance](https://docs.scipy.org/doc/scipy/reference/generated/scipy.spatial.distance.cosine.html)

Mathematical background for normalized dot-product similarity
between time-shifted signal segments.

### [3] Local Maxima and Peak Detection
[SciPy — find_peaks](https://docs.scipy.org/doc/scipy/reference/generated/scipy.signal.find_peaks.html)

Background for local maximum detection and the influence
of noise. The project uses its own implementation.

### [4] Phase Folding
[Astropy — Time Series Folding](https://docs.astropy.org/en/stable/timeseries/analysis.html#folding)

Background for folding observations onto a common period.
The project also averages corresponding positions across cycles.
