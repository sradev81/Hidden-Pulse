## Finding the Hidden Pulse — Final Results

I started this project because of a practical problem, which I encountered while editing my last project,a music videos clips for a symphony of five parts.<br> I noticed that some cuts felt natural at nearly regular intervals, even when there was no obvious drum hit or other strong musical event.

My goal was to investigate whether a mathematical method could detect this hidden timing structure and suggest useful editing points.

I tested the same method on three different compositions, without changing the detection algorithm or its parameters.

| Composition | Musical Meter | Detected Period | Practical Result |
|---|---|---|---|
| Earth | 4/4 | 2.60 s | Close alignment with manually selected cuts |
| Fire | 4/4 | 2.55 s | Useful rhythm after a phase adjustment of approximately 7–8 frames |
| Aether | 5/4 | 2.85 s | Useful rhythm after a phase adjustment of approximately 6 frames |

<br>

### What I Found

The method detected repeated timing patterns in all three compositions, including Aether, which has a different musical meter.

For Earth, eight representative manual cuts were compared quantitatively with the generated points. <br>The mean absolute difference was **1.5 frames**, or **62.5 milliseconds** at 24 fps.

For Fire and Aether, the detected periods produced useful editing grids, but their initial timing required small manual phase adjustments.

An important observation is that **finding the repetition period does not automatically mean finding the best exact moment for a cut**.

<br>

### Limitations

The current model detects periodicity in the local signal level. It does not distinguish individual musical instruments or reliably detect every strong musical attack.

This limitation was especially noticeable in Fire, where the music contains aggressive rhythmic events.

Also, a mathematically suggested point is not necessarily the best artistic decision.<br> Sometimes I intentionally place a cut a few frames earlier or later because of the visual content.

The practical comparisons are limited, so these results cannot establish universal accuracy for other compositions.

<br>

### Final Thoughts

The original question was whether mathematics could help me find natural editing points that I struggle sometimes to hear clearly between different instruments.

**My experiments suggest that it can.**

The method provided useful timing grids for three contrasting compositions, without requiring a different algorithm for each one.

It is not an automatic video editor (YET), and it does not replace my artistic decisions.<br> But it can give me a starting point instead of searching for every cut through repeated trial/error.

For me, that is the most useful result of this project. And I am so excited of it, becouse it will save me time and days of struggling.

After the project, I will continue developing the model by combining periodicity with the detection of stronger musical attacks for better performance<br> and accurancy, while keeping the final editing decisions under my control.
