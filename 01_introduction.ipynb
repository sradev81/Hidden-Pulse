## 1. Introduction

As a filmmaker and creator of music videos, my work often requires choosing cut points that feel natural with the rhythm and structure of the music.

In music with strong transients, such as pronounced drum hits or sharp attacks, or with Vocals, where the pacing is driven by high energy, suitable cut points may be relatively easy to identify. In smoother musical passages, however, natural editing points may follow a less obvious periodic pulse that is not represented by a single strong sound event.

My last project was of that kind. A symphony from 5 parts - The five elements by Adrian Velner. First 3 parts were more dynamic and was kind of easy. But I spent days getting through the first half of a part IV - Earth, cutting again and again, replaying tiny fragments many times to find the right editing points, so the changing of the video to stay naturally. In the middle of the song, I realized that several consecutive clips had almost the same duration: 2 seconds and 20 frames (02:20 at 24 fps, approximately 2.83 s).

I decided to test this observation. I cut the next clip to the same duration without listening for a strong hit, without looking at the waveform, and without searching for an obvious peak. It fit naturally. I continued using the same interval and finished almost the entire second half of the video in about two hours, using this interval almost all the time, with few exceptions.

<div style="height: 4px;"></div>

With this project, I will investigate whether simple mathematical analysis of an audio signal can detect these underlying rhythmic patterns and identify useful candidate points for video editing, potentially saving significant editing time and nerves.


&nbsp;

### Research Question

Can a general mathematical method detect candidates, for cut points that align with my manually selected (I guess this should be the period of the song, which I am loosing between the different instrumentals, becouse I am not a musiciаn), across both strongly percussive and smooth musical passages?

<div style="height: 4px;"></div>

### Project Scope

The project focuses on mathematical signal analysis.

The initial approach will examine:
- changes in local signal energy;
- repeated temporal patterns;
- periodicity in the audio signal;
- candidate cut points derived from these patterns.

<div style="height: 4px;"></div>

The detected points will be later compared with my manually selected edit points from the original finished video projects.
